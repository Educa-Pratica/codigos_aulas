```python
from langchain.chat_models import init_chat_model

provider = "openai"                        # template base da OpenAI
model    = "speakleash/Bielik-Minitron-7B-v3.0-Instruct-GGUF:Q4_K_M"  # nome do modelo
config   = provider + ":" + model

params = {
    "model": config,
    "api_key": "",
    "temperature": 0.3,
    "max_tokens": 1500,
    # "extra_body": {"repeat_penalty": 1.15, "min_p": 0.05},
    "reasoning": {"effort": "low", "summary": "auto"},
    "base_url": "http://127.0.0.1:8080/v1",  # API endpoint
}

llm = init_chat_model(**params)
response = llm.invoke("escreva um poema sobre python")
```
